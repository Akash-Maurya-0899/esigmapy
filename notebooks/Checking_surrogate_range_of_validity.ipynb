```python
import os
os.environ["OMP_NUM_THREADS"] = "1" # Required for efficient surrogate evaluation by avoiding multi-threading overheads
import numpy as np
import matplotlib.pyplot as plt
import esigmapy.esigmaSur as esigmasur

# Configuring some plot settings
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Computer Modern"],
    "font.size": 12
    })

sur_data_dir = "../esigmapy/esigmaSur/data"
sur_data_dir = os.path.abspath(sur_data_dir)
os.environ['ESIGMASUR_DATA_PATH'] = sur_data_dir # Point to the directory where surrogate data is stored (esigmapy/esigmaSur/data)
```

```python
M = 10.
reference_eccentricity = 0.431 # reference eccentricity
# reference_mean_anomaly = 60 * np.pi/180. # reference mean anomaly

distance = 400. # source luminosity distance (in Mpc)
inclination = 30 * np.pi/180. # orbital inclination with line-of-sight

delta_t = 1/2**12 # time grid-spacing (in s)
t_start = -0.1 # Waveform starting time (in s). t=0 corresponds to the end of waveform, so the starting time should be a negative real number

N = int(np.sqrt(1000))

rng = np.random.default_rng(seed=37)
q_vals = rng.uniform(1, 6, size=N)             
l_vals = rng.uniform(0, 2 * np.pi, size=N)
# e_vals = rng.uniform(5.e-7, 0.431, size=N)        

q_edges = [1., 7.]
l_edges = [0., np.pi, 2*np.pi]
# e_edges = [5.e-7, 0.431]

q_array = np.r_[q_edges, q_vals]
l_array = np.r_[l_edges, l_vals]
# e_array = np.r_[e_edges, e_vals]


for q in q_array:
    for reference_mean_anomaly in l_array:
    # for reference_eccentricity in e_array:
        m1 = q*M/(1+q)
        m2 = M/(1+q)
        try:
            modes = esigmasur.get_inspiral_esigmasur_modes(
                mass1=m1,
                mass2=m2,
                reference_eccentricity=reference_eccentricity,
                reference_mean_anomaly=reference_mean_anomaly,
                delta_t=delta_t, 
                t_start=t_start,
                distance=distance)
        except Exception as e:
            print(f"Failed at q={q}, l={reference_mean_anomaly} with exception\n {e}" )
```
